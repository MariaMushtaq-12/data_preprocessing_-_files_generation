generate merged dataframe into a geojson

```python
import geopandas as gpd
import pandas as pd

# Load shapefile
gdf = gpd.read_file('/content/sample_data/data/Pakistan.shp')

# Load CSV
df = pd.read_csv('/content/sample_data/data/tehsil_Copy.csv')

# Strip, lower, and clean the 'ADM3_EN' and 'Tehsil' columns
gdf['ADM3_EN'] = gdf['ADM3_EN'].str.strip().str.lower()
df['Tehsil'] = df['Tehsil'].str.strip().str.lower()

# Merge the dataframes
merged = gdf.merge(df, left_on='ADM3_EN', right_on='Tehsil', how='left')

# Specify the output file path for GeoJSON (change to your desired path)
output_geojson_path = '/content/sample_data/data/merged_geojson.geojson'

# Save the merged GeoDataFrame as a GeoJSON file
merged.to_file(output_geojson_path, driver='GeoJSON')

print(f"Merged GeoJSON file created and saved successfully at: {output_geojson_path}")
```

identify which tehsils in your CSV file were not merged by checking for NaN values in the resulting merged GeoDataFrame. If a tehsil from the CSV does not match any tehsil in the shapefile, it will have NaN values for the columns that came from the shapefile.


```python
# Get the tehsils in the CSV that were not merged (i.e., no matching tehsil in the shapefile)
unmerged_tehsils = df[~df['Tehsil'].isin(merged['Tehsil'])]

# Alternatively, you can check for NaN values in the merged DataFrame
unmerged_tehsils_nan = merged[merged['ADM3_EN'].isna()]

# Print the results
print("Tehsils in the CSV that were not merged:")
print(unmerged_tehsils[['Tehsil']])

print("\nTehsils that resulted in NaN values after merging:")
print(unmerged_tehsils_nan[['Tehsil']])
```

To generate a CSV file containing the tehsils from the CSV that did not merge, you can modify the previous code to export the unmerged tehsils as a CSV file.

```python
# Get the tehsils in the CSV that were not merged (i.e., no matching tehsil in the shapefile)
unmerged_tehsils = df[~df['Tehsil'].isin(merged['Tehsil'])]

# Specify the output file path for the unmerged tehsils CSV
output_csv_path = '/content/sample_data/data/unmerged_tehsils.csv'

# Save the unmerged tehsils to a CSV file
unmerged_tehsils[['Tehsil']].to_csv(output_csv_path, index=False)

print(f"CSV file of unmerged tehsils created and saved successfully at: {output_csv_path}")
```
